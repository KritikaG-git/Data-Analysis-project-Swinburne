### Task 3 - Test & Train Data Preparation

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
clean_data = pd.read_csv('clean_data.csv')
```

```python
# Drop columns that are not needed
new_data = clean_data.drop(['Observer'], axis=1)
# 'Observer' column was dropped as it was not adding anything to this task.

new_data.to_csv('new_data.csv', index=False)
print(new_data.head())
```

```python
new_data
```

```python
new_data.dtypes
```

```python
new_data['Location'] = new_data['Location'].astype('category').cat.codes
#changing categorical columns to numerical
#East Carlburgh: 0
#Ramirezview: 2 
#East Carloston: 1(these values were matched with new_data table)
```

```python
#changing values of Y and N to 1 and 0
new_data['Crop Damage'] = new_data['Crop Damage'].map({'Y': 1.0, 'N': 0.0})
```

```python
new_data['Heat Wave'] = new_data['Heat Wave'].map({'Y': 1.0, 'N': 0.0})
```

```python
new_data['Cold Wave'] = new_data['Cold Wave'].map({'Y': 1.0, 'N': 0.0})
```

```python
new_data['Dry Spell'] = new_data['Dry Spell'].map({'Y': 1.0, 'N': 0.0})
```

```python
new_data['Wet Spell'] = new_data['Wet Spell'].map({'Y': 1.0, 'N': 0.0})
```

```python
new_data['Irrigation'] = new_data['Irrigation'].map({'Y': 1.0, 'N': 0.0})
```

```python
new_data['Crop Type'] = new_data['Crop Type'].astype('category').cat.codes
#canola: 0
#wheat: 2
#maize :1
```

```python
mean_temperature = new_data['Avg Min Temp °C'].mean()
new_data['Avg Min Temp °C'].fillna(mean_temperature, inplace=True)
```

```python
new_data["Avg Min Temp °C"] = new_data["Avg Min Temp °C"].round(2)
```

```python
new_data
```

```python
new_data.to_csv('new_data.csv', index=False)
```

```python
from sklearn.model_selection import train_test_split
#splitting of data into training and testing
```

```python
#Dataset (a) - Yield as the target variable
# Drop the 'Yield' column from X
X_a = new_data.drop('Yield', axis=1)
# Set 'Yield' as the target y
y_a = new_data['Yield'] 

# Split the data into training and testing sets (80:20)
a_X_train, a_X_test, a_y_train, a_y_test = train_test_split(X_a, y_a, test_size=0.2, random_state=42)

# Saving the csv files
a_X_train.to_csv('a_X_train.csv', index=False)
a_X_test.to_csv('a_X_test.csv', index=False)
a_y_train.to_csv('a_y_train.csv', index=False)
a_y_test.to_csv('a_y_test.csv', index=False)

# Dataset (b) - Crop Damage as the target variable
# Drop the 'Crop Damage' column from X
X_b = new_data.drop('Crop Damage', axis=1) 
# Set 'Crop Damage' as the target y
y_b = new_data['Crop Damage'] 

# Split the data into training and testing sets (80:20)
b_X_train, b_X_test, b_y_train, b_y_test = train_test_split(X_b, y_b, test_size=0.2, random_state=42)

# Saving the CSV files
b_X_train.to_csv('b_X_train.csv', index=False)
b_X_test.to_csv('b_X_test.csv', index=False)
b_y_train.to_csv('b_y_train.csv', index=False)
b_y_test.to_csv('b_y_test.csv', index=False)
```
