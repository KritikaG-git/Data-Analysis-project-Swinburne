```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
##loading the csv files
X_train = pd.read_csv('a_X_train.csv')
y_train = pd.read_csv('a_y_train.csv')
X_test = pd.read_csv('a_X_test.csv')
y_test = pd.read_csv('a_y_test.csv')
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor

# Create the Linear Regression model
linear_regression = LinearRegression()

# Create the Decision Tree model
decision_tree = DecisionTreeRegressor()

# Fit the models on the training data
linear_regression.fit(X_train, y_train)
decision_tree.fit(X_train, y_train)

# Make predictions on the testing data
linear_regression_preds = linear_regression.predict(X_test)
decision_tree_preds = decision_tree.predict(X_test)
```

```python
#working with five differnt configurations:
```

```python
# Configuration 1:
#LR with default configuration:
linear_regression_default = LinearRegression()
linear_regression_default.fit(X_train, y_train)
linear_regression_preds_default = linear_regression_default.predict(X_test)
```

```python
#Configuration 2:
## LR with regualtion ridge
from sklearn.linear_model import Ridge
ridge_regression = Ridge(alpha=0.5)
ridge_regression.fit(X_train, y_train)
ridge_regression_preds = ridge_regression.predict(X_test)
```

```python
#Configuration 4:
#DT with max depth

decision_tree_max_depth = DecisionTreeRegressor(max_depth=3)
decision_tree_max_depth.fit(X_train, y_train)
decision_tree_preds_max_depth = decision_tree_max_depth.predict(X_test)
```

```python
#Configuration 5:
#DT min samples leaf

decision_tree_min_samples_leaf = DecisionTreeRegressor(min_samples_leaf=5)
decision_tree_min_samples_leaf.fit(X_train, y_train)
decision_tree_preds_min_samples_leaf = decision_tree_min_samples_leaf.predict(X_test)
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# Define the algorithm + configuration combinations
combinations = [
    {"Algorithm": "Linear Regression", "Configuration": "Default"},
    {"Algorithm": "Linear Regression", "Configuration": "With Regularization"},
    {"Algorithm": "Decision Tree", "Configuration": "Default"},
    {"Algorithm": "Decision Tree", "Configuration": "With Max Depth"},
    {"Algorithm": "Decision Tree", "Configuration": "With Min Samples Leaf"}
]

#Storing the assessment results
results_df = pd.DataFrame(columns=["Algorithm", "Configuration", "MSE", "RMSE", "MAE", "R-squared"])

# Assessment of all configurations 
for combination in combinations:
    algorithm = combination["Algorithm"]
    configuration = combination["Configuration"]
    
    # Train the model
    if algorithm == "Linear Regression":
        model = LinearRegression()
    elif algorithm == "Decision Tree":
        model = DecisionTreeRegressor()
    
    # Apply configuration if applicable
    if configuration == "With Regularization":
        # Apply regularization to the linear regression model
        model = Ridge(alpha=0.5)
    elif configuration == "With Max Depth":
        # Set maximum depth for the decision tree
        model = DecisionTreeRegressor(max_depth=3)
    elif configuration == "With Min Samples Leaf":
        # Set minimum samples per leaf for the decision tree
        model = DecisionTreeRegressor(min_samples_leaf=5)
    
    # fitting the model on the training data
    model.fit(X_train, y_train)
    
    # Evaluating the model on the test data
    y_pred = model.predict(X_test)
    
    #mse:mean sqaured error
    #rmse: root mean squared error
    #mae: mean absolute error
    #less the values of these better is the model and vice versa for r squared value
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    mae = mean_absolute_error(y_test, y_pred)
    r2 = r2_score(y_test, y_pred)
    
    #adding results to dataframe we created
    results_df = results_df.append({
        "Algorithm": algorithm,
        "Configuration": configuration,
        "MSE": mse,
        "RMSE": rmse,
        "MAE": mae,
        "R-squared": r2
    }, ignore_index=True)


print(results_df)
```
