```python
# loading the datafiles and importing packages
import pandas as pd
import numpy as np
```

```python
#first datafile
data_1_env = pd.read_csv('data_1_env.csv')
# as the table was provided for description of columns, required data cleaning and conversion is done accordingly.

#converting the temperature to celsius
data_1_env['Avg Min Temp *C'] = (data_1_env['Avg Min Temp *F'] - 32) * 5 / 9

# renaming the column
data_1_env = data_1_env.rename(columns={'Avg Min Temp *C': 'Avg Min Temp °C'})

# removing the Fahrenheit column as we already converted the values
data_1_env = data_1_env.drop(columns=['Avg Min Temp *F'])
data_1_env
```

```python
#converting rainfall from inches to mm
data_1_env['Yr Rain mm'] = data_1_env['Yr Rain inches'] * 25.4

# renaming the column
data_1_env = data_1_env.rename(columns={'Yr Rain mm': 'Yr Rain mm'})

# dropping the inches column as we already converted the values
data_1_env = data_1_env.drop(columns=['Yr Rain inches'])
print(data_1_env)
```
